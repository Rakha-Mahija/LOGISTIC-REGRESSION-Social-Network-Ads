## Step 1: Import necessary libraries
Start by importing the required modules for data manipulation, modeling, and evaluation.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, classification_report
from sklearn.preprocessing import StandardScaler
```

## **Step 2: Prepare your data**

For this example, you will create a synthetic dataset using NumPy. The X array will contain the feature (input) data, and the y array will contain the binary target (output) data (0s and 1s).

```python
import kagglehub
import os

# 1. Download dataset terbaru (mengembalikan path folder)
path = kagglehub.dataset_download("dragonheir/logistic-regression")
#print("Path ke folder dataset:", path)

# 2. Cari nama file di dalam folder tersebut
files = os.listdir(path)
#print("File yang ditemukan:", files)

# 3. Gabungkan path folder dengan nama file
path = "/kaggle/input/logistic-regression"
os.listdir(path)
file_path = path + "/Social_Network_Ads.csv"

# 4. pisahkan training set dan testing set
df = pd.read_csv(file_path)
X = df[['Age', 'EstimatedSalary']]
y = df['Purchased']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

## **Step 3: Create and train the model**

Next, create an instance of the LogisticRegression class and train it using the `fit()` method on your training data.

```python
# Create a logistic regression model instance
model = LogisticRegression(solver='liblinear', random_state=0)

# Train the model with the training data
model.fit(X_train, y_train)
```

**Interpreting the coefficients:**

After training, you can inspect the model's coefficients `(coef_)` and intercept `(intercept_)`. The coefficient represents the change in the log-odds of the outcome for a one-unit change in the predictor variable.


```python
# Get the model's coefficients and intercept
print(f"Intercept: {model.intercept_}")
print(f"Coefficients: {model.coef_}")

# You can convert the coefficients to odds ratios by exponentiating them
odds_ratio = np.exp(model.coef_)
print(f"Odds Ratio: {odds_ratio}")
# An odds ratio of 4.03, for instance, means that for every 1mm increase in tumor size,
# the odds of it being cancerous increase by 4x.
```

## **Step 4: Make predictions**

Use the trained model to make predictions on the test data. The `predict()` method returns the predicted class (0 or 1), while `predict_proba()` returns the probabilities of each class.

```python
# Predict the class labels for the test set
y_pred = model.predict(X_test)
print(f"Predicted labels: {y_pred}")

# Predict the probability of each class
y_pred_proba = model.predict_proba(X_test)
print(f"Predicted probabilities:\n {y_pred_proba}")
```

## **Step 5: Evaluate the model**

Evaluate your model's performance using metrics like accuracy, a confusion matrix, and a classification report.

```python
# Calculate the accuracy of the model on the test set
accuracy = model.score(X_test, y_test)
print(f"Model accuracy: {accuracy:.2f}")

# Display the confusion matrix
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(cm)
# The confusion matrix shows:
# [True Negatives, False Positives]
# [False Negatives, True Positives]

# Display a more comprehensive classification report
report = classification_report(y_test, y_pred)
print("Classification Report:")
print(report)
```
